```python
activities = activity.filter(
    target_chembl_id = "CHEMBL230"
)

print(len(activities))
```

```python
activity_record = activities[0]

print(activity_record)
```

```python
# Check for records where the measurement is IC50

ic50_activities = activity.filter(
    target_chembl_id="CHEMBL230",
    standard_type="IC50"
)

print(len(ic50_activities))
```

```python
# Inspect the amount of records/dict used in respect to the IC50 values
```

```python
Counter(a["standard_relation"] for a in ic50_activities)
```

```python
# Checking what the records with none standard value is like
for a in ic50_activities[:10]:
      if a["standard_relation"] == None:
          print(a)
    
```

```python
#  The records with only standard_relation == "=" will be considered
```

```python
exact_ic50 = [
    a for a in ic50_activities
    if a["standard_relation"] == "="
]
```

```python
len(exact_ic50)
```

```python
print(exact_ic50[1009])
```

```python
# Check for the variation in standard units
Counter(a["standard_units"] for a in exact_ic50)
```

```python
# We will use the records which have their standard_units == "nm"
```

```python
curated_ic50 = [
    a for a in exact_ic50
    if a["standard_units"] == "nM"
]
```

```python
len(curated_ic50)
```

```python
# Next big questions - How many unique molecules are represented by those 6,393 records?
```

```python
# Lets put the molcules ids in a set {} because it stores only unique values
molecule_ids = {
    a["molecule_chembl_id"] 
    for a in curated_ic50
}
```

```python
len(molecule_ids)
```

```python
# Count and get the amount of times each molecule_id is duplicated
record_count = Counter(
    a["molecule_chembl_id"]
    for a in curated_ic50
)
```

```python
# Check the number of molecules with 1,2,3.......etc molecule id

distribution = Counter(record_count.values())
print(distribution)
```

```python
# Calculate how many records are left over (duplicate records) after 1 of each record has been taken
```

```python
duplicate_records = sum(
    count - 1
    for count in record_count.values()
    if count > 1
)

print(duplicate_records)
```

```python
# Check how many molecules have multiple records (more than 1)
molecules_with_multiple_records = sum (
    count > 1
    for count in record_count.values()
)
```

```python
print(molecules_with_multiple_records)
```
