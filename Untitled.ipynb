```python
target = new_client.target
```

```python
activity = new_client.activity
```

```python
results = target.search("COX-2")
```

```python
# results
```

```python
for result in results[:10]:
    print(
          result["target_chembl_id"],
          result["pref_name"],
          result["target_type"]
         )
```

```python
for target_id  in ["CHEMBL4321", "CHEMBL2977", "CHEMBL1293198"]:
    result =  target.get(target_id)

    print("\nTARGET :", target_id)
    print("NAME:", result["pref_name"])
    print("TYPE:", result["target_type"])
    print("ORGANISM:", result.get("organism") )
    # print("TARGET COMPONENTS", result.get("target_components"))
```
